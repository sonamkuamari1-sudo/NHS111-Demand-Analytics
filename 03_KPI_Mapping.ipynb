```python
import pandas as pd

# Load the cleaned NHS111 dataset
file_path = "../data/cleaned/NHS111_Cleaned.csv"

df = pd.read_csv(file_path)

print("Cleaned dataset loaded successfully.")
print("Dataset shape:", df.shape)
```

```python
# Inspect all unique NHS111 item/KPI codes

item_codes = sorted(df["ITEM_NUMBER"].unique())

print("Total unique ITEM_NUMBER codes:", len(item_codes))
print("\nITEM_NUMBER codes:")
print(item_codes)
```

```python
# Check data availability for each ITEM_NUMBER

item_summary = (
    df.groupby("ITEM_NUMBER")
      .agg(
          Total_Rows=("VALUE", "size"),
          Valid_Values=("VALUE", "count"),
          Missing_Values=("VALUE", lambda x: x.isna().sum())
      )
      .reset_index()
)

item_summary["Missing_Percent"] = (
    item_summary["Missing_Values"] / item_summary["Total_Rows"] * 100
).round(2)

item_summary = item_summary.sort_values(
    by="Valid_Values",
    ascending=False
)

print("KPI availability summary:")
print(item_summary.to_string(index=False))
```

```python
# Map key NHS111 item codes to KPI descriptions

kpi_mapping = {
    "A01": "Number of calls received",
    "A02": "Calls routed through IVR",
    "A03": "Number of answered calls",
    "A05": "External clinician calls to Clinical Assessment Service (CAS)",
    "A07": "Calls which originated from an external NHS 111 provider",
    "B01": "Number of calls answered within 60 seconds",
    "B02": "Number of calls abandoned",
    "B03": "Calls abandoned in 30 seconds or less",
    "B04": "Calls abandoned in over 30 seconds and up to and including 60 seconds",
    "B05": "Calls abandoned after 60 seconds"
}

# Create a KPI description column
df["KPI_DESCRIPTION"] = df["ITEM_NUMBER"].map(kpi_mapping)

print("KPI mapping applied successfully.")

print("\nMapped KPI counts:")
print(df["KPI_DESCRIPTION"].value_counts(dropna=False))
```

```python
# Save KPI-enriched dataset

output_path = "../data/cleaned/NHS111_KPI_Mapped.csv"

df.to_csv(output_path, index=False)

print("KPI-mapped dataset saved successfully.")
print("Saved to:", output_path)
print("Final shape:", df.shape)
```
