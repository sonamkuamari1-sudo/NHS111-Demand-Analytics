```python
import pandas as pd
import numpy as np

# Load raw NHS111 dataset
file_path = "../data/raw/IUCADC-Apr23-to-Mar24-Revised.csv"

df = pd.read_csv(file_path)

print("Raw dataset loaded successfully.")
print("Dataset shape:", df.shape)
```

```python
# Inspect the VALUE column before cleaning
print("VALUE data type:", df["VALUE"].dtype)

print("\nSample VALUE entries:")
print(df["VALUE"].head(15))

print("\nCount of '-' values:")
print((df["VALUE"] == "-").sum())
```

```python
# Clean the VALUE column

# Remove commas from numbers
df["VALUE"] = df["VALUE"].str.replace(",", "", regex=False)

# Replace "-" with missing values and convert to numeric
df["VALUE"] = pd.to_numeric(df["VALUE"], errors="coerce")

print("VALUE column cleaned.")
print("New data type:", df["VALUE"].dtype)
print("Missing values after conversion:", df["VALUE"].isna().sum())

print("\nSample cleaned values:")
print(df["VALUE"].head(15))
```

```python
# Clean and convert REPORTING_PERIOD to a proper date

df["REPORTING_PERIOD"] = (
    df["REPORTING_PERIOD"]
    .str.replace("IUCADC-MONTHLY-", "", regex=False)
)

df["REPORTING_PERIOD"] = pd.to_datetime(
    df["REPORTING_PERIOD"],
    format="%b-%Y"
)

print("REPORTING_PERIOD converted successfully.")
print("New data type:", df["REPORTING_PERIOD"].dtype)

print("\nReporting periods:")
print(sorted(df["REPORTING_PERIOD"].unique()))
```

```python
# Final validation after cleaning

print("Dataset shape:", df.shape)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nDate range:")
print("Start:", df["REPORTING_PERIOD"].min())
print("End:", df["REPORTING_PERIOD"].max())
```

```python
# Save the cleaned dataset

output_path = "../data/cleaned/NHS111_Cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully.")
print("Saved to:", output_path)
print("Final shape:", df.shape)
```
